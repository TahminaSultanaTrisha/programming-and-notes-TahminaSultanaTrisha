_What statement introduces a conditional statement in Python?_

The statement that introduces a conditional statement in Python is: if.                                           Example:

```python
x = 5
if x > 0:
    print("x is positive")
```

_And how is this used to handle two possible situations? Three? Four?_

*Two possible situations: Use if and else:*

```python
x = 5
if x > 0:
    print("x is positive")
else:
    print("x is zero or negative")
```

*Three possible situations: Use if, elif, and else:*

```python
x = 0
if x > 0:
    print("x is positive")
elif x < 0:
    print("x is negative")
else:
    print("x is zero")
```

*Four or more situations: Use multiple elif branches:*

```python
score = 85
if score >= 90:
    print("Grade A")
elif score >= 80:
    print("Grade B")
elif score >= 70:
    print("Grade C")
else:
    print("Grade D or lower")
```

_And what structure allows for an EAFP approach?_

The structure that allows for an EAFP (“Easier to Ask for Forgiveness than Permission”) approach in Python is the try…except block.  Example:


```python
try:
    value = int(input("Enter a number: "))
    print("You entered:", value)
except ValueError:
    print("That was not a valid number!")
```

```python


user_input = input("Enter distance in miles: ")

# First check: is the input a number?
if user_input.replace('.', '', 1).isdigit():  # allows decimal point
    miles = float(user_input)

    # Second check: is the value non-negative?
    if miles >= 0:
        kilometres = miles * 1.60934
        print(f"{miles} miles is equal to {kilometres:.2f} kilometres.")
    else:
        print("Error: Distance cannot be negative.")
else:
    print("Error: Invalid input. Please enter a number.")
```

_Suppose a kindly teacher has some sweets to share between a class of students (it can happen). Write a program below that takes twp inputs - the number of sweets and the number of students (both integers), and displays how many sweets each student gets (everyone gets the same), and how many will be left over for the teacher._

_Code the Happy Path first, in the box below._

```python


sweets = int(input("Enter the number of sweets: "))
students = int(input("Enter the number of students: "))

each_gets = sweets // students  # integer division
left_over = sweets % students  # remainder

print(f"Each student gets {each_gets} sweets.")
print(f"The teacher gets {left_over} sweets left over.")
```

_Now for some LBYL. We can't have negative values for the sweets or students. Zero would make sense, though. So add code now to detect these error cases and display a message. Can you get a different message for each case?_

_For full marks here (!), the program should really not ask for the number of students if the number of sweets is invalid._

```python


sweets_input = input("Enter the number of sweets: ")

# First, check if sweets input is a valid integer
if sweets_input.isdigit():
    sweets = int(sweets_input)

    # Check if sweets is negative
    if sweets < 0:
        print("Error: Number of sweets cannot be negative.")
    else:
        students_input = input("Enter the number of students: ")

        # Check if students input is a valid integer
        if students_input.isdigit():
            students = int(students_input)

            if students < 0:
                print("Error: Number of students cannot be negative.")
            else:
                # Everything is valid — perform division
                if students == 0:
                    print("No students — the teacher keeps all the sweets!")
                else:
                    each_gets = sweets // students
                    left_over = sweets % students
                    print(f"Each student gets {each_gets} sweets.")
                    print(f"The teacher gets {left_over} sweets left over.")
        else:
            print("Error: Invalid number of students. Please enter a whole number.")
else:
    print("Error: Invalid number of sweets. Please enter a whole number.")
```

_You can probably see by now how error-checking code can obscure what's really going on. There are still other errors possible here - what happens if the user enters a value that cannot be converted into an integer? Add some EAFP code to trap this (it will be enough to catch the Exception once, from either of the places it could be thrown)._

```python


try:
    sweets = int(input("Enter the number of sweets: "))
    if sweets < 0:
        print("Error: Number of sweets cannot be negative.")
    else:
        students = int(input("Enter the number of students: "))
        if students < 0:
            print("Error: Number of students cannot be negative.")
        elif students == 0:
            print("No students — the teacher keeps all the sweets!")
        else:
            each_gets = sweets // students
            left_over = sweets % students
            print(f"Each student gets {each_gets} sweets.")
            print(f"The teacher gets {left_over} sweets left over.")
except ValueError:
    print("Error: Invalid input. Please enter whole numbers only.")
```

_There's still another error. 0 sweets is fine, even though it would just be a sad day. But 0 students will make the program crash when it tries to divide by zero. We could use EAFP or LBYL here, but EAFP will be neater, and will keep the code out of the way. After all, trying to divide something between no-one is not very likely. If you try to enter a zero you will see it's a `ZeroDivisionError` that needs to be caught. Add that in (it's just two more lines) below._

```python


try:
    sweets = int(input("Enter the number of sweets: "))
    if sweets < 0:
        print("Error: Number of sweets cannot be negative.")
    else:
        students = int(input("Enter the number of students: "))
        if students < 0:
            print("Error: Number of students cannot be negative.")
        elif students == 0:
            # we could LBYL-check here, but we'll let the ZeroDivisionError handler manage any accidental division-by-zero
            print("No students — the teacher keeps all the sweets!")
        else:
            each_gets = sweets // students
            left_over = sweets % students
            print(f"Each student gets {each_gets} sweets.")
            print(f"The teacher gets {left_over} sweets left over.")
except ValueError:
    print("Error: Invalid input. Please enter whole numbers only.")
except ZeroDivisionError:
    # two-line addition: catch division-by-zero if it ever occurs
    print("Error: Cannot divide by zero (no students)!")
```

_Below, write code that would take two inputs, which are marks on a test, scored on a scale of 0 to 100. Both is worth 50%, so the overall mark is just the average. Your code should display the overall grade (`round` the average)._

_**Start with the Happy Path.** _Then spot and add code for the various possible errors._ _Add more code blocks to the Notebook if you want._ Add more Markdown blocks to help you work out the problem if it helps._

```python


mark1 = float(input("Enter the first mark (0–100): "))
mark2 = float(input("Enter the second mark (0–100): "))

average = (mark1 + mark2) / 2
print(f"The overall mark is {round(average)}%.")
```

```python


try:
    mark1 = float(input("Enter the first mark (0–100): "))
    mark2 = float(input("Enter the second mark (0–100): "))

    # Validate range
    if not (0 <= mark1 <= 100):
        print("Error: First mark must be between 0 and 100.")
    elif not (0 <= mark2 <= 100):
        print("Error: Second mark must be between 0 and 100.")
    else:
        average = (mark1 + mark2) / 2
        print(f"The overall mark is {round(average)}%.")
except ValueError:
    print("Error: Please enter numeric values for both marks.")
```

```python


mark1_input = input("Enter the first mark (0–100): ")
mark2_input = input("Enter the second mark (0–100): ")

# Check if both inputs are numeric (allow decimals)
if mark1_input.replace('.', '', 1).isdigit() and mark2_input.replace('.', '', 1).isdigit():
    mark1 = float(mark1_input)
    mark2 = float(mark2_input)

    # Check valid range for both marks
    if not (0 <= mark1 <= 100):
        print("Error: First mark must be between 0 and 100.")
    elif not (0 <= mark2 <= 100):
        print("Error: Second mark must be between 0 and 100.")
    else:
        average = (mark1 + mark2) / 2
        print(f"The overall mark is {round(average)}%.")
else:
    print("Error: Invalid input. Please enter numeric values only.")
```

```python


try:
    bill = float(input("Enter the total bill amount: "))
    tendered = float(input("Enter the amount tendered: "))

    if bill < 0:
        print("Error: Bill amount cannot be negative.")
    elif tendered < 0:
        print("Error: Tendered amount cannot be negative.")
    elif tendered < bill:
        print(f"Error: Insufficient payment. You still owe ${bill - tendered:.2f}.")
    else:
        change = round(tendered - bill, 2)
        print(f"Change due: ${change:.2f}")
except ValueError:
    print("Error: Invalid input. Please enter valid numbers.")
```

_Finish this week by creating a program to show the best way to provide the change, in the smallest possible number of notes and coins.Your program should take as input the amount of change to be given, and then output how to make the change._

_We'll assume that the change will just be given in coins. Coins exist for £2, £1, 50p, 20p, 10p, 5p, 2p, 1p._

_Hint: Although the number entered will be floating point, it will probably be easiest to convert it into an integer (the amount of pennies). Ignore the issue of more than two decimal places._

_Another Hint: If you find yourself writing a program with about 10 `if` statements you should pause and think about whether there is a better way. (There is.)_

```python


change = float(input("Enter the amount of change (£): "))

# convert pounds to pence
pennies = int(round(change * 100))

coins = [200, 100, 50, 20, 10, 5, 2, 1]
coin_names = ["£2", "£1", "50p", "20p", "10p", "5p", "2p", "1p"]

print(f"Change to give: £{change:.2f}")
for value, name in zip(coins, coin_names):
    count = pennies // value
    pennies %= value
    if count > 0:
        print(f"{count} × {name}")
```

```python


try:
    change = float(input("Enter the amount of change (£): "))

    if change < 0:
        print("Error: Change cannot be negative.")
    else:
        pennies = int(round(change * 100))  # convert to pence

        coins = [200, 100, 50, 20, 10, 5, 2, 1]
        coin_names = ["£2", "£1", "50p", "20p", "10p", "5p", "2p", "1p"]

        print(f"Change to give: £{change:.2f}")
        for value, name in zip(coins, coin_names):
            count = pennies // value
            pennies %= value
            if count > 0:
                print(f"{count} × {name}")
except ValueError:
    print("Error: Please enter a valid number for the change amount.")
```
