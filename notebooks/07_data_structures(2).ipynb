_Suppose you wanted to implement a queue system. This needs to add new elements at the end of the queue, and remove elements of the front. Is this a list, tuple, set, or dictionary?_

If we want to build a queue in Python, go with a list—or better yet, use `collections.deque` from the standard library.

So, why skip the others?
- Tuple: Nope. It’s locked down; you can’t add or remove stuff.
- Set: Not happening. The order gets all mixed up, so you lose the “first in, first out” idea.
- Dictionary: It’s for key-value pairs, not for lining things up one after another.

Now, why choose a list or a deque?
- Lists let you add things at the end with `.append()`, and you can take from the front with `.pop(0)`.
- Deques are even better. Adding or removing from either end is super fast—no waiting around for Python to shuffle things.

Here’s what it looks like:

from collections import deque

queue = deque()
queue.append("first")
queue.append("second")
print(queue.popleft())

_Suppose you wanted to implement an invoice system. It needs to take records of a purchase from a database, and print the lines on an invoice for a customer. Is this a list, tuple, set, or dictionary? Or maybe two of them?_

We’re building an invoice system, so we have to keep track of a few things. For the customer details : like name, address, and so on—it just makes sense to use a dictionary. It’s all key-value pairs, really straightforward.

When it comes to line items, things get a bit different. Each invoice can have several items, and the order matters. So, a list works best here. Each line item goes in as its own dictionary with fields like product, quantity, and price.

In short, we’ll use a dictionary for customer info and invoice details, and a list of dictionaries for the line items.

Here’s the basic setup: one dictionary holds the invoice number, date, and all the customer info. Then there’s a list, and each entry in that list is another dictionary—one for each line item. Simple and organized.

````python

customer = {
    "name": "John Smith",
    "address": "123 Main Street",
    "city": "New York"
}

invoice_items = [
    {"product": "Laptop", "price": 1200, "quantity": 1},
    {"product": "Mouse", "price": 25, "quantity": 2},
    {"product": "Keyboard", "price": 80, "quantity": 1}
]

def create_invoice(customer_info, items):
    print("INVOICE")
    print(f"Customer: {customer_info['name']}")
    print(f"Address: {customer_info['address']}, {customer_info['city']}")
    print()
    print("Items:")

    total = 0
    for item in items:
        item_total = item['price'] * item['quantity']
        total += item_total
        print(f"{item['product']} - ${item['price']} x {item['quantity']} = ${item_total}")

    print()
    print(f"Total Amount: ${total}")

create_invoice(customer, invoice_items)
````

_Write a program that displays three **unique** random numbers between 0 and 10._

_Hint: There is a built-in way to do this, but it's obscure. Use a loop, and a set here._

````python
import random

def get_unique_random_numbers():
    numbers = set()
    while len(numbers) < 3:
        n = random.randint(0, 10)
        numbers.add(n)
    return numbers

random_nums = get_unique_random_numbers()

print("Three unique random numbers:", random_nums)
````

_Rewrite the code above using a list._

_This is not that much more complex, but it shows why it's not always easiest to reach for a list. Think about your data!_

````python
import random

def get_unique_random_numbers_list():
    numbers = []
    while len(numbers) < 3:
        num = random.randint(0, 10)
        if num not in numbers:
            numbers.append(num)
    return numbers

random_nums = get_unique_random_numbers_list()

print("Three unique random numbers:", random_nums)
````

_Now a program that prompts the user to enter a collection of numbers (terminated by just pressing Enter), and then displays
the **list** in decreasing numerical order. For simplicity, assume a Happy Path._

````python
def sort_numbers_decreasing():
    numbers = []

    print("Type in your numbers one at a time. Just hit Enter when you’re done:")
    while True:
        user_input = input()
        if user_input == "":
            break
        numbers.append(float(user_input))

    numbers.sort(reverse=True)

    print("Here they are, biggest to smallest:", numbers)

sort_numbers_decreasing()
````

_How about the longest word in a sentence? You'll need to ignore punctuation characters. (For bonus marks, handle the possibility that there is more than one word of the longest length!)_

_Hint: You can sort a list by length ..._

_Another Hint: See below._

        sentence.replace(symbol, '')


````python
from string import punctuation

def find_longest_words():
    sentence = input("Enter a sentence: ")

    for symbol in punctuation:
        sentence = sentence.replace(symbol, '')

    words = sentence.split()
    if not words:

        print("No words found!")
        return

    max_length = max(len(word) for word in words)
    longest_words = [word for word in words if len(word) == max_length]

    print(f"Longest word(s) (length {max_length}): {longest_words}")

find_longest_words()
````

_See https://en.wikipedia.org/wiki/Sieve_of_Eratosthenes for a simple way to find Prime numbers. Use it to print a list of all the primes less than 100. Don't worry about too many optimisations!_

_Start with this list:_

    numbers = [True for n in range(101)]


````python

numbers = [True for n in range(101)]

numbers[0] = False
numbers[1] = False

for current_num in range(2, 101):
    if numbers[current_num]:
        multiple = current_num * 2
        while multiple < 101:
            numbers[multiple] = False
            multiple += current_num

prime_numbers = [num for num, is_prime in enumerate(numbers) if is_prime]
print("Prime numbers less than 100:")
print(prime_numbers)
````

_Use a list to implement a simple queue. Customers join a queue at the end, and leave from the front. You will need functions to print the current queue, add a customer, remove a customer, and display the current queue length. Use strings to represent customers._

_Add a short program to test the queue. Remember error conditions - for example, a customer cannot leave if the queue is empty!_


````python

class SimpleQueue:
    def __init__(self):
        self.queue = []

    def print_queue(self):
        if not self.queue:
            print("Queue is currently empty")
        else:
            print("Current queue:", " -> ".join(self.queue))

    def add_customer(self, customer):
        if not isinstance(customer, str):
            print("Error: Customer must be a string")
            return False

        self.queue.append(customer)
        print(f"Added customer '{customer}' to the queue")
        return True

    def remove_customer(self):
        if not self.queue:
            print("Error: Cannot remove customer - queue is empty!")
            return None

        customer = self.queue.pop(0)
        print(f"Removed customer '{customer}' from the queue")
        return customer

    def queue_length(self):
        length = len(self.queue)
        print(f"Queue length: {length}")
        return length

def test_queue():
    print("=== Simple Queue Test Program ===\n")

    queue = SimpleQueue()

    print("Test 1: Initial state")
    queue.print_queue()
    queue.queue_length()
    print()

    print("Test 2: Adding customers")
    queue.add_customer("Alice")
    queue.add_customer("Bob")
    queue.add_customer("Charlie")
    queue.print_queue()
    queue.queue_length()
    print()

    print("Test 3: Removing customers")
    queue.remove_customer()
    queue.print_queue()
    queue.queue_length()
    print()

    print("Test 4: Removing more customers")
    queue.remove_customer()
    queue.remove_customer()
    queue.print_queue()
    print()

    print("Test 5: Removing from empty queue")
    queue.remove_customer()
    print()

    print("Test 6: Adding new customers")
    queue.add_customer("Diana")
    queue.add_customer("Eve")
    queue.print_queue()
    queue.queue_length()
    print()

    print("Test 7: Invalid input")
    queue.add_customer(123)
    queue.print_queue()

if __name__ == "__main__":
    test_queue()
````

## Reflection

Python provides a basic collection of data structures. There are more available, but these cover all the common applications. The trick in writing a program efficiently is often to spot which one is needed. So:

- If you need to sort it, it's probably a list.
- If the values are different types, that's a tuple.
- A set is good when values need to be unique.
- And a dictionary is for key-value pairs.

Remember that you can convert between these. So the usual way to remove duplicates from a list would be to convert to a set, then back again. If you really need to sort a tuple, convert it to a list, sort, and convert back. And so on.

For some others in the Standard Library, see https://docs.python.org/3/library/collections.html
