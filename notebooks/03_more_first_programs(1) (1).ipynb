_What purpose does a blank line serve in a Python program?_



A **blank line** in Python helps to **improve readability** by visually separating sections of code.
It does **not affect how the program runs**, but makes it easier for humans to read and understand.

**Example:**

- Without blank lines

````python
name = "Alex"
age = 25
print(name)
print(age)
````

- With blank lines (more readable)

````python
name = "Alex"
age = 25

print(name)
print(age)
````

_Python programmers often refer to `PEP-8`, a much revered document. What is it? (Google is your friend here.)_



**PEP 8** is the **Python Enhancement Proposal 8**, a document that provides **guidelines and best practices** for writing **clean, readable, and consistent Python code**.

**Key points of PEP 8 include:**
- Naming conventions (e.g., lower snake case for variables)
- Indentation and spacing around operators
- Maximum line length (usually 79 characters)
- Blank lines to separate code sections
- Guidelines for comments and docstrings

**Why it matters:**
Following PEP 8 makes your code **easier for others (and yourself) to read**, and is widely adopted in the Python community.

**Example:**
```python
# PEP 8 compliant
user_name = "Alex"
age = 25

print(f"{user_name} is {age} years old.")


_Suppose you write a program, and the person sat next to you writes the same program. Assuming both work, would they be exactly the same? Why (not)?_



Even if both programs work correctly and produce the same result, they would **not necessarily be exactly the same**.

**Reasons:**
- Different variable names (e.g., `user_name` vs `name`)
- Different formatting or spacing
- Different ways to solve the problem (e.g., using a `for` loop vs a `while` loop)
- Different comments or structure

 **Key idea:**
Python allows multiple ways to solve the same problem. As long as the **logic and output are correct**, variations in style or approach are normal.

_Suppose you purchase a 10 inch pizza. How many square *centimetres* of pizza do you have? Write a program below to find out. Remember that 10 inches is the diameter of the pizza._

````python

diameter_in = 10
diameter_cm = diameter_in * 2.54
radius_cm = diameter_cm / 2
import math
area_cm2 = math.pi * radius_cm**2

print(f"The area of a {diameter_in}-inch pizza is {area_cm2:.2f} cm².")
````

_A 10 inch pizza costs £10.99. A 12 inch pizza costs £13.99. Which is the best buy in terms of pizza area for your pound? Write a program
that prompts for the size and cost, and prints the cost per square cm._

_Hint: It is probably easier to work in pence per square cm. Even then, the answers will be small numbers._

````python

import math
diameter_in = float(input("Enter pizza diameter in inches: "))
cost_pounds = float(input("Enter pizza cost in pounds: "))

cost_pence = cost_pounds * 100

diameter_cm = diameter_in * 2.54
radius_cm = diameter_cm / 2
area_cm2 = math.pi * radius_cm**2
cost_per_cm2 = cost_pence / area_cm2

print(f"Cost per square cm: {cost_per_cm2:.2f} pence/cm²")
````

_Peaches, on the other hand, are packed into boxes of 24 fruit. Any remaining after this are packed into smaller punnets of 4. Adapt the program above to handle peaches, and this more complex packing strategy._

````python

peaches_per_box = 24
peaches_per_punnet = 4

total_peaches = int(input("Enter the total number of peaches: "))

full_boxes = total_peaches // peaches_per_box
remaining_peaches = total_peaches % peaches_per_box

full_punnets = remaining_peaches // peaches_per_punnet
leftover_peaches = remaining_peaches % peaches_per_punnet

print(f"Full boxes of 24 peaches: {full_boxes}")
print(f"Full punnets of 4 peaches: {full_punnets}")
print(f"Leftover peaches: {leftover_peaches}")
````
